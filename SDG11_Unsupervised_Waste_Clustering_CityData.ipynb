This notebook applies **unsupervised learning (K-Means Clustering)** to analyze waste management efficiency across cities, using your provided dataset `city_level_data_0_0.csv`.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score
from sklearn.decomposition import PCA

# 1. Load Dataset
file_path = '/mnt/data/city_level_data_0_0.csv'
df = pd.read_csv(file_path)
print('Data Preview:')
display(df.head())
print('\nColumns available:', df.columns.tolist())

# 2. Data Preprocessing (update features based on your dataset columns)
numeric_df = df.select_dtypes(include=[np.number]).dropna()
scaler = StandardScaler()
X_scaled = scaler.fit_transform(numeric_df)
X_train, X_test = train_test_split(X_scaled, test_size=0.2, random_state=42)

# 3. Train K-Means Model
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X_train)
clusters_train = kmeans.labels_
clusters_test = kmeans.predict(X_test)

# 4. Evaluate Clustering Performance
silhouette = silhouette_score(X_train, clusters_train)
calinski = calinski_harabasz_score(X_train, clusters_train)
davies = davies_bouldin_score(X_train, clusters_train)
print(f'Silhouette Score: {silhouette:.3f}')
print(f'Calinski-Harabasz Index: {calinski:.3f}')
print(f'Davies-Bouldin Index: {davies:.3f}')

# 5. Visualize Clusters with PCA
pca = PCA(n_components=2)
X_2d = pca.fit_transform(X_train)
plt.figure(figsize=(8,6))
plt.scatter(X_2d[:,0], X_2d[:,1], c=clusters_train, cmap='viridis', alpha=0.7)
plt.title('K-Means Clusters (City-Level Data)')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.colorbar(label='Cluster')
plt.show()

# 6. Cluster Profiling
numeric_df['cluster'] = kmeans.labels_
cluster_profiles = numeric_df.groupby('cluster').mean()
print('\nCluster Profiles:')
display(cluster_profiles)
```
